```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('train_and_test2.csv')
print(df.shape)
df.head(10)
```

```python
## Step 1: Understand the dataset
```

```python
df.info()
```

```python
## Step 2: Basic statistics
```

```python
df.describe()
```

```python
## Step 3: Check missing values
```

```python
df.isnull().sum()
```

```python
## Step 4: Survival count
```

```python
df['2urvived'].value_counts()
```

```python
sns.countplot(x='2urvived', hue='Sex', data=df)
plt.title('Survival by gender')
plt.show()
```

```python
## Step 5: Survival by gender
```

```python
sns.countplot(x='2urvived', hue='Sex', data=df)
plt.title('Survival by Gender (0=No, 1=Yes)')
plt.xlabel('Survived')
plt.show()
```

```python
## Step 6: Age distribution
```

```python
sns.histplot(df['Age'], bins=30, kde=True)
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.show()
```

```python
## Step 6: Survival by passenger class
```

```python
sns.countplot(x='Pclass', hue='2urvived', data=df)
plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.show()
```

```python
## Step 8: Fare vs Age coloured by survival
```

```python
sns.scatterplot(x='Age', y='Fare', hue='2urvived', data=df, alpha=0.6)
plt.title('Fare vs Age by Survival')
plt.show()
```
